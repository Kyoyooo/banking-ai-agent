Chạy lệnh Pinggy để public cổng 11434 (Ollama)

```python
# 1. Cài đặt thư viện Pinggy
!pip install pinggy -q

import pinggy
import time

# 2. Khởi tạo Tunnel cho Ollama (Cổng 11434)
tunnel_ollama = pinggy.start_tunnel(forwardto="localhost:11434")

# 4. In ra đường link
print(f"Ollama API: {tunnel_ollama.urls[0]}")

print("Đang giữ kết nối... ")

# 5. Vòng lặp duy trì kết nối cho Colab cell
try:
    while True:
        time.sleep(1)
except KeyboardInterrupt:
    print("\nĐã đóng kết nối Pinggy.")
```
